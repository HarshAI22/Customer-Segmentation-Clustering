```python
# Cell 1
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.preprocessing import StandardScaler
from sklearn.cluster import KMeans, DBSCAN
from sklearn.metrics import silhouette_score
import joblib
import warnings
warnings.filterwarnings('ignore')
```

```python
# Cell 2
from sklearn.datasets import make_blobs
import io

# Mall Customers data manually create kar rahe hain
np.random.seed(42)
n = 200
income   = np.concatenate([np.random.normal(20,5,40), np.random.normal(60,10,60),
                            np.random.normal(90,8,40), np.random.normal(25,6,30),
                            np.random.normal(85,7,30)])
spending = np.concatenate([np.random.normal(20,5,40), np.random.normal(50,10,60),
                            np.random.normal(80,8,40), np.random.normal(75,7,30),
                            np.random.normal(20,5,30)])
gender   = np.random.choice(['Male','Female'], n)
age      = np.random.randint(18, 70, n)

df = pd.DataFrame({
    'CustomerID': range(1, n+1),
    'Gender': gender,
    'Age': age,
    'Income': np.clip(income, 10, 140).astype(int),
    'Spending': np.clip(spending, 1, 100).astype(int)
})
df.head()
```

```python
# Cell 4
df.isnull().sum()
```

```python
fig, axes = plt.subplots(1, 3, figsize=(15, 4))
sns.histplot(df['Age'], ax=axes[0], color='steelblue', kde=True)
axes[0].set_title("Age Distribution")
sns.histplot(df['Income'], ax=axes[1], color='coral', kde=True)
axes[1].set_title("Income Distribution")
sns.histplot(df['Spending'], ax=axes[2], color='green', kde=True)
axes[2].set_title("Spending Score Distribution")
plt.tight_layout()
plt.show()
```

```python
# Cell 8
X = df[['Income', 'Spending']]
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)
```

```python
# Cell 9
inertia = []
K_range = range(2, 11)
for k in K_range:
    km = KMeans(n_clusters=k, random_state=42)
    km.fit(X_scaled)
    inertia.append(km.inertia_)

plt.figure(figsize=(8, 4))
plt.plot(K_range, inertia, 'bo-')
plt.xlabel("Number of Clusters (K)")
plt.ylabel("Inertia")
plt.title("Elbow Method")
plt.show()
```

```python
kmeans = KMeans(n_clusters=5, random_state=42)
df['KMeans_Cluster'] = kmeans.fit_predict(X_scaled)
print("Silhouette Score (KMeans):", silhouette_score(X_scaled, df['KMeans_Cluster']))
```

```python
plt.figure(figsize=(8, 5))
sns.scatterplot(data=df, x='Income', y='Spending', hue='KMeans_Cluster', palette='tab10', s=80)
plt.title("KMeans Clustering — Income vs Spending")
plt.show()
```

```python
dbscan = DBSCAN(eps=0.5, min_samples=5)
df['DBSCAN_Cluster'] = dbscan.fit_predict(X_scaled)
print("Unique DBSCAN labels:", df['DBSCAN_Cluster'].unique())
print("Outliers (-1):", (df['DBSCAN_Cluster'] == -1).sum())
```

```python
plt.figure(figsize=(8, 5))
sns.scatterplot(data=df, x='Income', y='Spending', hue='DBSCAN_Cluster', palette='tab10', s=80)
plt.title("DBSCAN Clustering — Outliers in Black (-1)")
plt.show()
```

```python
# Cell 14
joblib.dump(kmeans, "kmeans_model.pkl")
joblib.dump(scaler, "scaler.pkl")
print("Saved!")
```
